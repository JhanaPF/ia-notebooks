Count images with 3 and 1 channels

```python
grey_count = 0
color_count = 0
```

```python
import cv2
import os
import random

def load_images_from_folder(folder_path, image_files):
    global grey_count, color_count
    for image_file in image_files:
        image_path = os.path.join(folder_path, image_file)
        image = cv2.imread(image_path)
        print(image.shape)
        if len(image.shape) == 2:  # Image en niveaux de gris
            grey_count+=1
        else:
            color_count+=1

TRAIN_DATA_PATH = 'datasets/chest_Xray/train'
VAL_DATA_PATH = 'datasets/chest_Xray/val'
TEST_DATA_PATH = 'datasets/chest_Xray/test'
data_paths = [TRAIN_DATA_PATH, VAL_DATA_PATH, TEST_DATA_PATH]

for data_path in data_paths:

    # Get Normal Images
    n_paths = os.listdir(os.path.join(data_path, 'NORMAL'))
    # Sort by name
    n_paths.sort()
    # remove the hidden files
    n_paths = [f for f in n_paths if f.endswith('.jpeg')]

    # Get Pneumonia Images
    p_paths = os.listdir(os.path.join(data_path, 'PNEUMONIA')) 
    
    # Sort by name
    p_paths.sort()
    # remove the hidden files
    p_paths = [f for f in p_paths if f.endswith('.jpeg')]

    # Iterate through the NORMAL folder
    load_images_from_folder(os.path.join(data_path, 'NORMAL'), n_paths)

    # Iterate through the PNEUMONIA folder
    load_images_from_folder(os.path.join(data_path, 'PNEUMONIA'), p_paths)

```

```python
print(grey_count, color_count)
```

```python
from PIL import Image
import matplotlib.pyplot as plt

labels = ['RGB', 'Noir et Blanc']
counts = [color_count, grey_count]

plt.figure(figsize=(8, 6))
plt.bar(labels, counts, color=['blue', 'gray'])
plt.xlabel('Type d\'image')
plt.ylabel('Nombre d\'images')
plt.title('Répartition des images RGB et Noir et Blanc')
plt.show()
```
